```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("https://raw.githubusercontent.com/majidramzi/data-analytics-ml/refs/heads/main/Palmer-Archipelago-Penguins/penguins_size.csv", encoding='latin-1')
df = df.dropna()
```

```python
# 1. Box plots: Body mass and Flipper length grouped by species
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(data=df, x='species', y='body_mass_g', palette='Set2', ax=axes[0])
axes[0].set_title('Body Mass Distribution by Species')

sns.boxplot(data=df, x='species', y='flipper_length_mm', palette='Set3', ax=axes[1])
axes[1].set_title('Flipper Length Distribution by Species')

plt.tight_layout()
plt.show()
```

```python
# 2. Scatter plot: Bill length vs. Bill depth (sized by body mass)
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x='culmen_length_mm', y='culmen_depth_mm', hue='species',
                size='body_mass_g', sizes=(20, 200), palette='Set1', alpha=0.7)
plt.title('Bill Length vs. Bill Depth (Sized by Body Mass)')
plt.xlabel('Bill Length (mm)')
plt.ylabel('Bill Depth (mm)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

The Palmer Penguins dataset displays distinct physical traits for each species.

The box plots of body mass and flipper length demonstrate that Gentoo penguins are significantly larger and heavier than both Adelie and Chinstrap species, with almost no overlap in size distribution.

The scatter plot mapping bill length against bill depth further separates the species into three clear clusters.

While Adelie and Chinstrap penguins share similar weights, Chinstraps have noticeably longer bills, making it easy to classify the birds based on these specific measurements.
