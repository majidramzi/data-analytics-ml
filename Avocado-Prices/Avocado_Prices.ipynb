```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/majidramzi/data-analytics-ml/refs/heads/main/Avocado-Prices/avocado.csv", encoding='latin-1')
df.head()
```

```python
# Cell left empty as the verification is now integrated above.
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Adjusting the overall image size
plt.figure(figsize=(12, 5))

# Chart 1: Box plot of price by avocado type
plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='type', y='AveragePrice', palette='Set2')
plt.title('Average Price: Organic vs Conventional')

# Chart 2: Boxplot of price by year
plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='year', y='AveragePrice', palette='Set3')
plt.title('Average Price Distribution by Year')

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Converting the date column to a standard format for plotting a line chart
df['Date'] = pd.to_datetime(df['Date'])

plt.figure(figsize=(15, 5))

# Chart 1: Box plot of price by type
plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='type', y='AveragePrice', palette='Set2')
plt.title('Average Price: Organic vs Conventional')

# Second chart: Price boxplot by year
plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='year', y='AveragePrice', palette='Set3')
plt.title('Average Price by Year')

plt.tight_layout()
plt.show()

# Chart 3: Sales volume over time
plt.figure(figsize=(12, 5))
sns.lineplot(data=df, x='Date', y='Total Volume')
plt.title('Total Volume Over Time')
plt.show()
```

The analysis of the Avocado dataset reveals a clear price difference based on the farming method.

Organic avocados consistently show a higher median price and a wider price distribution compared to conventional ones, which remain tightly grouped at a lower cost.

Additionally, tracking the total volume over time highlights distinct seasonal demand spikes, showing how purchasing behavior fluctuates throughout the year.
