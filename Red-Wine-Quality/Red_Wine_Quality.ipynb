```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("https://raw.githubusercontent.com/majidramzi/data-analytics-ml/refs/heads/main/Red-Wine-Quality/winequality-red.csv")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Box plot: Alcohol content grouped by Quality
sns.boxplot(data=df, x='quality', y='alcohol', palette='Set2', ax=axes[0])
axes[0].set_title('Alcohol Content by Wine Quality')
axes[0].set_xlabel('Quality Score')
axes[0].set_ylabel('Alcohol (%)')

# Box plot: Volatile Acidity grouped by Quality
sns.boxplot(data=df, x='quality', y='volatile acidity', palette='Set3', ax=axes[1])
axes[1].set_title('Volatile Acidity by Wine Quality')
axes[1].set_xlabel('Quality Score')
axes[1].set_ylabel('Volatile Acidity')

plt.tight_layout()
plt.show()
```

The Red Wine Quality dataset demonstrates strong correlations between specific chemical properties and the final quality score.

The box plots reveal two distinct trends.

First, higher quality wines (scores 7 and 8) consistently contain higher median alcohol percentages compared to lower-rated wines.

Conversely, volatile acidity shows a strong negative correlation with quality; as the wine quality score increases, the concentration of volatile acidity noticeably decreases.

This makes alcohol and volatile acidity strong predictive features for determining premium wine.
