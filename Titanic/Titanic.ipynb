```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read the training dataset from GitHub repository
df_titanic = pd.read_csv("https://raw.githubusercontent.com/majidramzi/data-analytics-ml/refs/heads/main/Titanic/train.csv", encoding='latin-1')

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Box plot: Age distribution grouped by Passenger Class
sns.boxplot(data=df_titanic, x='Pclass', y='Age', palette='Set2', ax=axes[0])
axes[0].set_title('Age Distribution by Passenger Class')
axes[0].set_xlabel('Passenger Class (1 = 1st, 2 = 2nd, 3 = 3rd)')

# 2. Box plot: Fare distribution grouped by Survival status
sns.boxplot(data=df_titanic, x='Survived', y='Fare', palette='Set3', ax=axes[1])
axes[1].set_title('Ticket Fare Distribution by Survival Status')
axes[1].set_xlabel('Survived (0 = No, 1 = Yes)')

plt.tight_layout()
plt.show()
```

The Titanic dataset provides an excellent example of data skewness and the impact of high-value outliers.

The box plot analyzing age across passenger classes reveals that first-class passengers were generally older, with a higher median age compared to those in third class.

Furthermore, the box plot mapping ticket fares against survival status displays a heavy right skew.

The significant number of extreme high-value outliers among the survivors visually demonstrates that passengers who paid premium fares had a substantially higher survival rate.
